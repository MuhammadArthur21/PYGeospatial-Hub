# Shapely Introduction
Basic geometric operations using Shapely
## Creating Geometries

```python
from shapely.geometry import Point, LineString, Polygon

point = Point(106.8, -6.2)
line = LineString([(106.7, -6.3), (106.9, -6.1)])
polygon = Polygon([(106.7, -6.3), (106.9, -6.3), (106.9, -6.1), (106.7, -6.1)])

print(f"Point: {point}")
print(f"Area: {polygon.area}")
```

```python
buffered = polygon.buffer(0.01)
print(f"Buffered area: {buffered.area}")

intersection = polygon.intersection(point.buffer(0.005))
print(f"Intersection: {intersection}")
```
