# Raster Data Processing with Rasterio
Reading and analyzing raster datasets

```python
import rasterio
import numpy as np

# Create sample raster
data = np.random.randint(0, 255, (100, 100), dtype='uint8')

with rasterio.open('sample.tif', 'w', driver='GTiff',
                    height=100, width=100, count=1,
                    dtype='uint8', crs='EPSG:4326') as dst:
    dst.write(data, 1)

print(f'Raster shape: {data.shape}')
print(f'Min: {data.min()}, Max: {data.max()}, Mean: {data.mean():.1f}')
```

```python
with rasterio.open('sample.tif') as src:
    band = src.read(1)
    print(f'CRS: {src.crs}')
    print(f'Bounds: {src.bounds}')
    print(f'Statistics: min={band.min()}, max={band.max()}')
```
