# GeoPandas Spatial Analysis
Working with spatial data using GeoPandas

```python
import geopandas as gpd

world = gpd.read_file(gpd.datasets.get_path('naturalearth_lowres'))
print(f'Countries: {len(world)}')
print(world.columns.tolist())
print(world[['name', 'continent', 'pop_est']].head())
```

```python
asia = world[world['continent'] == 'Asia']
area = asia.geometry.area / 1e6

top5 = asia.iloc[area.argsort().values[::-1][:5]]
print('Top 5 largest Asian countries:')
print(top5[['name', 'pop_est']])
```

```python
import matplotlib.pyplot as plt
fig, ax = plt.subplots(figsize=(8, 6))
asia.plot(ax=ax, color='#99AD7A', edgecolor='white')
ax.set_title('Asian Countries')
plt.show()
```
